```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ratings_df = pd.read_csv(
    "../data/ml-100k/u.data",
    sep="\t",
    names=["user_id","item_id","rating","timestamp"]
)

ratings_df.head()
```

```python
ratings_df.shape
```

```python
ratings_df.info()
```

```python
num_users = ratings_df["user_id"].nunique()
num_movies = ratings_df["item_id"].nunique()

print("Users:", num_users)
print("Movies:", num_movies)
```

```python
ratings_df["rating"].value_counts().sort_index()
```

```python
ratings_df["rating"].hist(bins=5)

plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.show()
```

```python
ratings_per_user = (
    ratings_df
    .groupby("user_id")
    .size()
)

ratings_per_user.describe()
```

```python
ratings_per_user.hist(bins=50)

plt.title("Ratings Per User")
plt.show()
```

```python
ratings_per_movie = (
    ratings_df
    .groupby("item_id")
    .size()
)

ratings_per_movie.describe()
```

```python
ratings_per_movie.hist(bins=50)

plt.title("Ratings Per Movie")
plt.show()
```

```python
total_possible = num_users * num_movies

actual_ratings = len(ratings_df)

sparsity = 1 - (actual_ratings / total_possible)

print(f"Sparsity: {sparsity:.4f}")
```
