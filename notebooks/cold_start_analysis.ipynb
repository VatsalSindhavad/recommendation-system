```python
import pandas as pd
```

```python
ratings_df = pd.read_csv(
    "../data/ml-100k/u.data",
    sep="\t",
    names=["user_id", "item_id", "rating", "timestamp"]
)
```

```python
user_activity = (
    ratings_df
    .groupby("user_id")
    .size()
)

user_activity.describe()
```

```python
cold_users = user_activity[
    user_activity <= 5
]

print("Cold Users:", len(cold_users))
```

```python
user_activity.hist(bins=50)
```

```python
def recommend_for_new_user(ratings_df, k=10):

    popular_items = (
        ratings_df
        .groupby("item_id")
        .size()
        .sort_values(ascending=False)
        .head(k)
    )

    return popular_items.index.tolist()
```

```python
recommend_for_new_user(ratings_df)
```
