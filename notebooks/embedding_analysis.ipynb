```python
from pathlib import Path

PROJECT_DIR = Path.cwd().resolve()
if PROJECT_DIR.name == "notebooks":
    PROJECT_DIR = PROJECT_DIR.parent

import sys
import torch
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA

sys.path.append(
    str(PROJECT_DIR / "src")
)

from model import NCF
```

```python
ratings_df = pd.read_csv(
    PROJECT_DIR / "data" / "ml-100k" / "u.data",
    sep="\t",
    names=["user_id","item_id","rating","timestamp"]
)

ratings_df["user_id"] = (
    ratings_df["user_id"]
    .astype("category")
    .cat.codes
)

ratings_df["item_id"] = (
    ratings_df["item_id"]
    .astype("category")
    .cat.codes
)

num_users = ratings_df["user_id"].nunique()
num_items = ratings_df["item_id"].nunique()
```

```python
model = NCF(
    num_users=num_users,
    num_items=num_items
)

model.load_state_dict(
    torch.load(
        PROJECT_DIR / "results" / "ncf_model.pth",
        map_location="cpu"
    )
)

model.eval()
```

```python
movie_embeddings = (
    model.item_embedding.weight
    .detach()
    .numpy()
)

movie_embeddings.shape
```

```python
pca = PCA(n_components=2)

movie_2d = pca.fit_transform(
    movie_embeddings
)

movie_2d.shape
```

```python
plt.figure(figsize=(12,8))

plt.scatter(
    movie_2d[:,0],
    movie_2d[:,1],
    alpha=0.5
)

plt.title(
    "Movie Embedding Space (PCA)"
)

plt.xlabel("Component 1")
plt.ylabel("Component 2")

plt.show()
```
